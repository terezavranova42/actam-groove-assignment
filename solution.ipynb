```python
from IPython.display import Audio
from drumkit import *

def play(wave, sample_rate=SR):
    return Audio(wave, rate=sample_rate, normalize=False, autoplay=False)
```

## Tempo and grid

```python
bpm = 71
step_s = 60 / bpm / 4   # one step = one sixteenth
```

## Patterns

One track per instrument. Replace this example with what you hear.

```python
tracksA = [("xx.....x..xx....", kick, step_s),
           ("..x...x.", snare, step_s*2),
           ("x.x.x.x.x.x.x.x.", closed_hat, step_s),
    ]

patternA = render_tracks(tracksA)


tracksB1 = [("x..x...x..xx...x", kick, step_s),
           ("..x...x.", snare, step_s*2),
           ("x.x.x.x.x.x.x.x.", closed_hat, step_s),
           ("x...", crash, step_s*4)]

tracksB2 = [(".x.x.x.x.x.x.x..", kick, step_s),
           ("x...x...x...x..x", snare, step_s),
           (".x...x...x...x..", open_hat, step_s)]

tracksB3 = [(".x...x...x...x..", kick, step_s),
           ("xxxx", snare, step_s*4),
           (".x...x...x...x..", open_hat, step_s)]

patternB = render_tracks(tracksB1) + render_tracks(tracksB2) + render_tracks(tracksB1) + render_tracks(tracksB3)
```

## Listen

```python
play(patternA*4 + patternB)
```
